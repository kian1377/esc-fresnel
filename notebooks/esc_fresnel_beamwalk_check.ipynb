```python
pointing = source_offset * M.as_per_lamD
print(f'Field = ', pointing)

ray_shifts = []
beam_shifts = []
optic_names = []

for i in range(len(wfs_on_axis)):
    optic_name = wfs_on_axis[i].location[6:]
    optic_name = wfs_on_axis[i].location[6:].lower().split(' ')[0]
    # print(name)
    if optic_name in list(beamwalk_matrices.keys()):
        print(i, optic_name)
        expected_shift_mm = beamwalk_matrices[optic_name].dot(pointing.to(u.mas)).to_value(u.mm)
        total_expected_shift_mm = np.sqrt( expected_shift_mm[0]**2 + expected_shift_mm[1]**2)
        # print(f'Expected shift: ( {expected_shift_mm[0]:.2e}, {expected_shift_mm[1]:.2e} )')
        print(f'Expected shift: {total_expected_shift_mm:.6f} mm')

        amp1 = wfs_on_axis[i].amplitude
        amp2 = wfs_offset[i].amplitude
        wf_pxscl_mm = wfs_on_axis[i].pixelscale.to_value(u.mm/u.pix)

        shift_pix, error, phasediff = phase_cross_correlation(
            ensure_np_array(amp1), 
            ensure_np_array(amp2),
            upsample_factor=1000,
        )

        shift_mm = np.flip(shift_pix) * wf_pxscl_mm
        total_shift_mm = np.sqrt( shift_mm[0]**2 + shift_mm[1]**2)
        # print(f'Measured shift: ( {shift_mm[0]:.2e}, {shift_mm[1]:.2e} )')
        print(f'Measured shift: {total_shift_mm:.6f} mm')

        ray_shifts.append(total_expected_shift_mm)
        beam_shifts.append(total_shift_mm)
        optic_names.append(optic_name)
        print()

ray_shifts = np.array(ray_shifts)
beam_shifts = np.array(beam_shifts)
```

```python
from tabulate import tabulate

headers = [
    'Optic\nName', 
    'Ray Shift From\nZemax Design Data [\u00b5m]', 
    'Beam Shift From\nFresnel Model Data [\u00b5m]',
    'Absolute\nDifference [\u00b5m]',
    'Percent\nError',
]

table_format_data = []
for i in range(len(optic_names)):
    if optic_names[i] in ['pupil_mask']: continue

    diff = np.abs(ray_shifts[i] - beam_shifts[i])
    pe = diff / ray_shifts[i] * 100
    table_row = [  
        optic_names[i].upper(), 
        ray_shifts[i] * 1000,
        beam_shifts[i] * 1000, 
        diff*1000, 
        pe, 
    ]

    table_format_data.append(table_row)
    

N = 16
table_1 = tabulate(
    table_format_data[:N], 
    headers=headers, 
    tablefmt="fancy_grid", 
    # colalign=('center', 'center', 'center', 'right', 'right'),
    colalign=['center']*5,
    # maxcolwidths=[8]*5,
    floatfmt=".3f",
)
print(table_1)

table_2 = tabulate(
    table_format_data[N:], 
    headers=headers, 
    tablefmt="fancy_grid", 
    # colalign=('center', 'center', 'center', 'right', 'right'),
    colalign=['center']*5,
    # maxcolwidths=[8]*5,
    floatfmt=".3f",
)
print(table_2)
```
